1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `Duck`

```python
#a
numbers=np.arange(1,101)
results = numbers.astype('U10')
print(numbers)
mickeys=np.where(numbers%3==0)
mouse=np.where(numbers%5==0)
mix=np.where(numbers%15==0)
results[mickeys[0]]='Mickey'
results[mouse[0]]='Mouse'
results[mix[0]]='MickeyMouse'
print(results)
#b
results=np.char.replace(results,'Mickey','Donald')
results=np.char.replace(results,'Mouse','Duck')
results=tuple(results)
print(results)
```

2\. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
def swap(x,y):
    z=x
    x=y
    y=z
    return x,y

def swapef(x,y):
    return y,x

x=[1,2,3]
y=[4,5,6]
print(swap(x,y))
print(swapef(x,y))
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
def distance(x,y):
    return np.sqrt((x[0]-y[0])**2+(x[1]-y[1])**2)
u=(0,3)
v=(4,0)
print(distance(u,v))
```

4\. Counting letters

Write a program to calculate the number of times each character occurs in a given string *s*. Ignore differneces in capitalization

```python
import numpy as np
import matplotlib.pyplot as plt

raw_chars = list('hello this is a test: kafdnanoanoucbenclanciohqeochlasncoisovcnalb')
string = np.char.lower(np.array(raw_chars))
letter, count = np.unique(string, return_counts=True)

# argsort returns the indices that would sort the array from Low -> High
sort_indices = np.argsort(count)

sorted_letters = letter[sort_indices]
sorted_counts = count[sort_indices]

print(np.column_stack((sorted_letters, sorted_counts)))

plt.bar(sorted_letters, sorted_counts)
plt.show()
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
def square(x):
    return x**2
def cube(x):
    return x**3
def sixth(x):
    return square(cube(x))

print(sixth(2))
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
def normalizarion(x):
    norm=np.sqrt(np.sum(np.array(x)**2))
    return x/norm
v=(1,3,5,3,3,5,5,3,2)
print(normalizarion(v))
```
