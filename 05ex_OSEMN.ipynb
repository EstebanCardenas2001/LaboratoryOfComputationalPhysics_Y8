4\. load the binary file named *credit_card.dat* and convert the data into the real credit-card number.
Each line correspond to a credit card number.
Each character is composed by 6 bit (even the space) and the last 4 bit are just a padding

**hint**: use the `chr()` function to convert a number to a char

```python
import numpy as np
def decode_credit_card_string(file_path):
    with open(file_path, 'rb') as file:
            raw_bytes = file.read()
    binary_string_raw = raw_bytes.decode('ascii')
    for i in range(len(binary_string_raw)//118):
        intervals=binary_string_raw[0+(119*i):118+(119*i)]
        card_number_raw = ""
        for j in range(19):
            start_bit = 6 * j
            end_bit = start_bit + 6
            six_bit_chunk = intervals[start_bit:end_bit]
            code = int(six_bit_chunk, 2)
            decoded_char=chr(code)
            card_number_raw += decoded_char
        print(card_number_raw)
    return card_number_raw
    
final_result = decode_credit_card_string('credit_card.dat')
print(final_result)
```
